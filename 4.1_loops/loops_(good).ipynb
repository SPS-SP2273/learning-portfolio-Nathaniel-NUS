# Interrupting the loop

### `break`

- The `break` statement will break out of the loop and terminate it

- Usually used together with `if` to allow us to terminate the loop during certain conditions

```python
for power in range(5):
    number = 10**power
    if number > 5000:
        break
    print(power, number)

x = 1
while x > 0:
    if x > 10:
        break
    x += 1 
    print(x)
```

### `continue` statement

- Skips the current iteration and moves on to the next

- Usually used together with `if` to allow us to skip an iteration during certain conditions

```python
for number in range(10):
    #Skip if the number is even (i.e. remainder is zero)
    if number % 2 == 0:
        continue
    print(number)
```

# List comprehension

- List comprehensions are shortcuts for **creating lists using loops** (instead of using for loop and append)

- Syntax `[<output> for <variable> in <iterable> if <condition>]`

- The iterable can be a list, string or range()

```python
x_list = [x for x in range(10) if x % 2 != 0]
print(x_list)

superhero_list = ["Iron Man", "Hulk", "Thor"]
y_list = [f"{name} is a Marvel superhero" for name in superhero_list]
print(y_list)

string_list = [x for x in 'impossible']
print(string_list)
```

### for with unpacking

- A for loop can have multiple loop variables to unpack higher dimensional lists/tuples

```python
Data = [("apple", 1), ("banana", 2), ("cherry", 3)]

for fruit, number in Data:
    number_squared = number ** 2
    print(f"{fruit}, {number_squared}")
```

- Remember, * operator can be used to unpack multiple items into a list

```python
items = [("a", 1, 2, 3), ("b", 4, 5)]

for first, *rest in items:
    print(first, rest, sep = ':')
```

### for with zip()

- zip() can be used to combine multiple lists. It returns an **iterator of tuples**, where the i-th tuple contains the i-th elements from each argument iterable

- Use a **for loop** to view the items in the new list. (Alternatively, list() and tuple() also works)

- If the lists are of different lengths, the list with shortest length will define the length of the combined list

```python
x = [1, 2, 3]
y = ["A", "B", "C"]
z = zip(x, y)
print(z)  #<zip object at 0x000001B30BB42C40. 

#Python does not process it until you use a for loop/ list()/ tuple()
for i in z:
    print(i)

#However, once you process it, Python forgets about it. z cannot be used anymore
#A way around this is to directly call: z = list(zip(x,y))

print(list(z))
print(tuple(z))
```

### Looping through dictionaries 

- Looping through the dictionary directly returns the keys


```python
letters_dict = {'a':1, 'b': 2, 'c': 3}
for i in letters_dict:
    y = letters_dict[i]
    print(y)
```

- Combine the for loop with keys(), items(), values() to have more functionality

```python
#Looping through keys

letters_dict = {'a':1, 'b': 2, 'c': 3}

for i in letters_dict.keys():
    print(i)
```

```python
#Looping through values

letters_dict = {'a':1, 'b': 2, 'c': 3}

for x in letters_dict.values():
    print(x)


#Alternatively, you can return the keys and use it to access values

for x in letters_dict.keys():
    print(letters_dict[x])
```

```python
#Looping through items

letters_dict = {'a':1, 'b': 2, 'c': 3}

#items() method will return each key:value pair as tuples in a list: dict_items([('a', 1), ('b', 2), ('c', 3)])
#Unpacking the items:
for i,y in letters_dict.items():  
    print(i,y)  # i = 1, y = 1. Not a nested loop, so i and y will increment together 
```
