# Path

- A path is just the location/address of a file/folder in the computer

- **Absolute path** is the complete location of a file or directory starting from the root of the file system. 

    - Windows uses `\` as path separators while macOS and Linux uses `/`. Hence, the **absolute path on Windows will not be recognisable by macOS or Linux, and vice versa**

```python
#Examples of absolute paths

#Windows
"C:\\Users\chammika\Desktop\data-01.txt"

#macOS (or Linux)
"/Users/chammika/Desktop/data-01.txt"
```

# Opening and Closing Files

- Syntax: `with open({file_name}, {'r'/'w'/'a'}) as alias`


- open() function is used to open the file. The with statement automatically closes the file once the code is ran (unclosed files can cause the file to be corrupted)

- `with` statement only opens files that exist 

```python
#Reading Data 

#'r' specifies that you only want to read. You cannot use write() if you specify 'r'
with open('Testing.txt', 'r') as file:
    file_content = file.read()

print(file_content)
```

```python
#(Over)writing) Data

#Likewise, 'w' specifies that you want to write. You cannot use read() if you specify 'w' 

text = "Hello I am adding this part using the write() function!!"
with open('Testing.txt', 'w') as file:
    file.write(text)

#If you want to read the file you have to open it separately
with open("Testing.txt", 'r') as file:
    file_content = file.read()
print(file_content)
```

```python
#Appending Data

#'a' specifies that you are opening the file to append

with open('Testing.txt', 'a') as file:
    for line in text.splitlines():
        file.write(line)

with open("Testing.txt", 'r') as file:
    file_content = file.read()
print(file_content)
```

# Folders

### Creating Folders

- Folders can be created using `os.mkdir(path)`. The path should be the address that the created folder should have

- Can be used in for loops to create a folder for iterables

```python
#E.g. Creating a new folder called Testing_Folder

os.mkdir("Testing_Folder")
```

```python
os.mkdir('people') #Making the folder called people

for person in ['John', 'Paul', 'Ringo']:
    path = os.path.join('people', person) #Adjust path to the subfolder for each person
    print(f'Creating {path}') 
    os.mkdir(path)  #Makes subfolder for each person

#if you do not adjust the path, Python will create main folders, not subfolders
```

---

### Checking for existence 

- Python will not let you move a file if the file already exist in the path specified; Python will not let you create a folder with a path that already exists 

- Folders/ files can be checked using `os.path.exists(path)` 

- Alternatively, try-except (with FileExistsError) can be used 

```python
#try-except 

for person in ['John', 'Paul', 'Ringo']:
    path = os.path.join('people', person)
    try:
        os.mkdir(path)
        print(f"Creating folder {person}")
    except FileExistsError: 
        print(f"This folder already exist!")
```

```python
#os.path.exist()

for person in ['John', 'Paul', 'Ringo']:
    path = os.path.join('people', person)
    if os.path.exists(path):
        print(f"This folder already exist!")
    else:
        os.mkdir(path)
        print(f"Creating folder {person}")        
```

### Copying Files

- Files can be copied using `shutil.copy(current_file_path, destination_file_path)`

```python
#Copying the SPS Logo Image into each person's folder

for person in ['John', 'Paul', 'Ringo']:
    path_to_destination = os.path.join('people', person) #folder path that it is moved to 
    shutil.copy('sp2273_logo.png', path_to_destination)
    print(f'Copied file to {path_to_destination}') 
```

### Moving Files

- Files can be moved using `shutil.move(current_file_path, destination_file_path)`

```python
for person in ['John', 'Paul', 'Ringo']:
    #Create folder imgs 
    path_to_imgs = os.path.join('people', person, 'imgs')
    if not os.path.exists(path_to_imgs):
        os.mkdir(path_to_imgs)
    
    #Move logo file
    current_path_of_logo = os.path.join('people', person, 'sp2273_logo.png')
    new_path_of_logo = os.path.join('people', person, 'imgs', 'sp2273_logo.png')

    shutil.move(current_path_of_logo, new_path_of_logo)
    print(f'Moved logo to {new_path_of_logo}')
```

# Listing Files

- `glob.glob()` function can be called to return a list of the specified files 

- `*` means anything 

- Each time you use `/`, it goes down into a subfolder

```python
#List all files in current folder 
glob.glob('*')
```

```python
#Files that match 'peo', followed by anything 
glob.glob('peo*')

#If you do not add the * -> youre asking to find folders with exactly this name
glob.glob('peo')
```

```python
#Files that match 'peo', followed by anything -> List anything inside the folders that start with 'peo'

glob.glob('peo*/*')
```

```python
#** and recursive = True gives the entire, detailed structure of the folder 

glob.glob('people/**', recursive=True)
```

```python
#Listing certain file types (E.g. png)

#'people folder' -> go through entire structure (** and recursive)-> anything with .png 
glob.glob('people/**/*.png', recursive=True)
```

```python
#To specify files only and not folders, you can use: 

#all files in the 'os-collab exercise subfolder' 
cairo_list = glob.glob('*/*cairo.*')
```

# Extracting file info 

- Often, file names, extensions or folders need to be extracted. This requires all the separators (\ . -) to be removed

- This can be done by splitting the string based on the separator: `string.split('separator')` 

- `os.path.sep` refers to either `/` or `\`

- Indexing must be used to get the part of the string you want

```python
path = r'people\Ringo\imgs\sp2273_logo.png'
filename = path.split(os.path.sep)[-1]
print(filename)
extension = filename.split('.')[-1]
print(extension)
```

```python
path = r'people\Ringo\imgs\sp2273_logo.png'

#Splitting filename from the rest
print(os.path.split(path))  #('people\\Ringo\\imgs', 'sp2273_logo.png')

#Splitting extension from the rest
print(os.path.splitext(path))  #('people\\Ringo\\imgs\\sp2273_logo', '.png')

#Show folder only
print(os.path.dirname(path))  #people\Ringo\imgs
```
