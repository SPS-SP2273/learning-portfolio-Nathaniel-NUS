# Basics

```python
#Cannot plot single points

x = 1 
y = 5
plt.plot(x,y)
plt.xlim([0,3])
plt.ylim([0,6])
plt.show()

```

```python
#Lists can be directly used in the plot function 

plt.plot([1,2,3],[4,5,6])

plt.show()

```

```python
#Lists can be used for data points of x and y. Ensure that they have the same number of elements 

x = [1,2,3,4,5]
y = [5,10,14,22,28]

plt.plot(x,y)
plt.show()
```

```python
#Arrays are better than lists so use them. Because you can do math operations directly on each element in arrays 

x = np.array([1,2,3,4,5])
y = np.array([5,10,14,22,28])

plt.plot(x,y)
plt.show()

```

```python
#Can use math eqns to determine y value

x = [0, np.pi/2, np.pi, 3*np.pi/2 , 2*np.pi]
y = np.sin(x)

plt.plot(x,y)
plt.show()

#Not a good plot, since we have too little points for the graph to be plotted accurately
```

```python
#Use the np.linspace function. Syntax: np.linspace(lower limit, Upper limit, Number of data points)

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x,y)
plt.show()
```

#### Adding Titles, labels and legends

```python
# Adding title
# Syntax: plt.title('string')

#can also change color, size, roation and opacity of the title

#If you are gonna type math equations, use raw string format r"$eqn$" so that Python does not treat \ as an escape sequence 

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x,y)
plt.title(r"$y=\sin x$", color = 'g', size = 50, alpha = .125, rotation = 180)
plt.show()

```

```python
# Adding Axis Labels 
# Syntax: plt.xlabel("string") and plt.ylabel("string")

#can also change color, size, roation and opacity of the title

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x,y)
plt.xlabel('x', color = "orange", alpha = .5, size = 20, rotation = 50)
plt.ylabel(r'$\sin x$')

plt.show()
```

```python
#Adding legend. 
#Syntax: plt.legend("string")

#Can specify location of legend and frame on/ off

#Cannot specify color. color will follow line color

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x,y, c= 'r')
plt.legend('y', frameon = False, loc = 'upper left')

plt.show()
```

```python
#Alternatively, 

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x,y, c= 'r', label = 'y')
plt.legend(frameon = False, loc = 'upper left') 

plt.show()
```

### Filling area 

```python
#Filling area under/above curve. 
# Syntax: plt.fill_between(x boundary, y boundary, color = ' ' )

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x,y)
plt.fill_between(x, y, color = 'r')
plt.show()

#You can control opacity by adding the argument alpha = .xxx 
# Alpha takes values from 0 - 1


plt.fill_between(x, y, color = 'r', alpha = .125)
plt.show()
```

```python
# Changes to line

#Any changes to the line itself will be passed as an argument to the plt.plot() function. Order does not matter

#Can change line color, line type, line opacity   


x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x,y, color = 'g', alpha = 1, linestyle = '--')


plt.fill_between(x, y, color = 'r', alpha = .125)
plt.show()



#Extra: You can specify colors using the Hex Code 
```

```python
#Adding markers to the line. Arguments are passed into plt.plot() function

#Marker type: marker = 'x' 
#Marker size: markersize = x or ms = x
#Color of the edge of marker: markeredgecolor = 'x' or mec = 'x' 
#Marker Color: markerfacecolor = 'x' or mfc 'x'

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x,y, marker = '*', ms = 10, mec = 'y', mfc = 'b')
plt.show()
```

png can have transparency but jpg cannot -> on mac png can get rid of background

zoom into png: pixelated

zoom into pdf: smooth line no matter how much u zoom -> quality remains same -> vector graphics -> pdf does not actually have the image, it has equations to draw the image -> each time u zoom it calculates what eqns to plot

# twinx()

```python
x = np.arange(10)
y2 = x**2 
sin_x = np.sin(x)

plt.plot(x,y2, label = r'$x^2$')
plt.plot(x,sin_x, label = r'$\sin x$')
plt.xlabel('x')
plt.ylabel('y')
plt.title(r'$y = x^2$')
plt.grid(alpha = .25, color = 'g')
plt.legend()
plt.show()

#The graph does not look very good 
```

```python
x = np.arange(10)
y2 = x**2 
sin_x = np.sin(x)

plt.plot(x,y2)
plt.ylabel(r'$y^2$', rotation = 0)

plt.twinx()  #Create new y axis 

plt.plot(x,sin_x, color = 'orange')  #Plots first graph, creates second y axis then plots second graph
plt.ylabel(r'$\sin x$', rotation = 0)

plt.xlabel('x')
plt.title(r'$y = x^2$')
plt.grid(alpha = .25, color = 'g')
plt.legend()
plt.show()


#y label on left side must be specified before twinx
#y label on right side must be specified after twinx
# same for legends, must have one bfr twinx and one after
```

# Error Bars

```python
#Syntax: plt.errorbar(x value, y value, x error, y error)

#errors must have same shape as their values 

#same error for all values 
x = np.arange(0,10)
y2 = x**2 


plt.errorbar(x,y2, label = r'$y^2$', xerr=.2, yerr=2)
plt.ylabel(r'$y^2$', rotation = 0)

plt.xlabel('x')
plt.grid(alpha = .25, color = 'g')
plt.legend()
plt.show()
```

```python
#Error as a percentage of its value 
x = np.arange(0,10)
y2 = x**2 


plt.errorbar(x,y2, label = r'$y^2$', xerr=.2, yerr=y2/10)
plt.ylabel(r'$y^2$', rotation = 0)

plt.xlabel('x')
plt.grid(alpha = .25, color = 'g')
plt.legend()
plt.show()
```

# PLotting Real Life Data

```python
url = "https://berkeley-earth-temperature.s3.us-west-1.amazonaws.com/Global/Complete_TAVG_daily.txt"
data = np.loadtxt(url, skiprows=23)

years = data[:,1]
anomaly = data[:,-1]

plt.plot(years, anomaly)
plt.hlines(y=0, xmin = years[0], xmax = years[-1], colors = 'red', linestyle = '--')
```
