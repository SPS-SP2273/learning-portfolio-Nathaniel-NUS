# Error Handling

### `assert` Statement

- Syntax: `assert {condition}, {message}`

- The assert statement returns an AssertationError and stops the program if the condition is False 

```python
x = 10
assert x >= 0, "x cannot take negative values"
print("This line will be printed if there are no errors")
```

```python
x = -3
assert x >= 0, "x cannot take negative values"
print("This line will be printed if there are no errors")
```

assert statements can be used in loops and functions as well

```python
def convert_celsius(temperature, target_scale):
    assert target_scale == "Kelvin" or "Fahrenheit", "Error, target scale can only be K or F"  #Note: if you put the "Kelvin" or "Fahrenheit" in parentheses, it will not work. ("Kelvin" or "Fahrenheit") will return "Kelvin", and the assert statement will only raise the error if the rtarget scale is "Kelvin"
    if target_scale == 'Kelvin':
        tempearture_kelvin = temperature + 273.15
        return f"{tempearture_kelvin} K"
    else:
        temperature_fahrenheit = (temperature * 1.8) + 32
        return f"{temperature_fahrenheit} F"

print(convert_celsius(20, 'Kelvin'))
print(convert_celsius(20, "Fahrenheit"))
print(convert_celsius(20, 2))
```

### `try-except`

- If left unhandled, exceptions will stop the program from running

- try and except is a way of handling these errors without causing the program to stop 

- The try block runs the code that may potentially give errors. The except block will run if Python detects the error 

```python
number = input("Please enter a number")
square = int(number)**2  #Converts the string to an integer. However this only works if the string is a numerical one   
print(f'The square of {number} is {square}!')

ValueError: invalid literal for int() with base 10: 'p' 
```

```python
try:
    number= input ("Please enter a number")
    square=int(number)**2
    print(f'The square of {number} is {square}!')
except:
    print("Please enter integers only!!")
```

### Positional, keyword and default arguments 

- Positional Arguments: Values are passed into the function in the **same order** that the parameter was defined 

- Keyword Arguments: Values passed are **specified by which parameter they belong to**. Order does not matter

- Default Arguments: Default arguments are specified in the function definition. If not arguments are passed when calling the function, the default values are used

```python
def side_by_side(a, b, c=42):
    return f'{a: 2d}|{b: 2d}|{c: 2d}'

print(side_by_side(1,2,3)) #3 Positional Arguments
print(side_by_side(c=3, b=1, a=2)) #3 Keyword Arguments 
print(side_by_side(1, b=2))  #1 positional, 1 keyword and 1 default
```

```python
#Some other examples

side_by_side(1, 2)           # Two positional, 1 default
## ' 1| 2| 42'
side_by_side(1, 2, 3)        # Three positional
## ' 1| 2| 3'
side_by_side(a=1, b=2)       # Two keyword, 1 default
## ' 1| 2| 42'
side_by_side(c=3, b=1, a=2)  # Three keyword
## ' 2| 1| 3'
side_by_side(1, c=3, b=2)    # One positional, 2 keyword
## ' 1| 2| 3'
side_by_side(1, b=2)         # One positional, 1 keyword, 1 default
## ' 1| 2| 42'
```

- Note: Positional arguments must always come first if not there would be errors

```python
side_by_side(b=2, 1, 3)  
# SyntaxError: positional argument follows keyword argument

def side_by_side(a=3, b, c):
    return f'{a: 2d}|{b: 2d}|{c: 2d}'

side_by_side(1, 2, 3)
# SyntaxError: parameter without a default follows parameter with a default
```

### Docstrings

- Docstrings are written in between ` ''' ` or `""" ` 

- Docstrings are used to **document what the code does inside the function**. If you call the help() function on the defined function, the documentation will be displayed 

```python
def my_function(a,b):
    """ 
    The function returns the product of a and b
    """
    return a * b

help(my_function)
```

### Unpacking

### Unpacking variables

- You can assign elements from lists or arrays into multiple variables in one line. This is called unpacking variables

```python
import numpy as np
#Unpacking from a list
x, y, z = [1,2,3]
print(x, y, z)

#Unpacking from an array
a, b, c = np.array([4,5,6])
print(a, b, c)
```

```python
#If the number of variables is less than the number of elements in a list, you can use the * operator 

# * operator assigns the remaning elements as a list to a variable with the * operator

x, y, *z = np.array([1,2,3,4,5])  #x and y will have the first two elements, and everything else is assigned to z as a list 
print(x, y, z)
```

```python
#The operator can be on any of the variables, and Python will adjust accordingly 

x, *y, z = np.array([1,2,3,4,5])  #x and z will have the first and last elements respectively, and everything else is assigned to y as a list 
print(x, y, z)
```
