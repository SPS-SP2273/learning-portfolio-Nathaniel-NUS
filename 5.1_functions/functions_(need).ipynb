```python
#from matplotlib import pyplot as plt
import random
import numpy as np
from matplotlib import pyplot as plt
```

```python
def straight_line(m,c):
    x_values = [x for x in range(0,10)]
    y_values = [m * x + c + random.random() for x in x_values]
    print(x_values)
    print(y_values)

straight_line(4,11)
```

```python
def numpy_straight_lines(m,c):
    x_values = np.linspace(0,100)
    y_values = m*x_values + c + 50*np.random.random()
```

```python
def write_to_data(m,c):
    x_values = np.linspace(0,10,100)
    y_values = m*x_values + c + 5*np.random.random()
    np.savetxt('hahah.txt', (x_values, y_values))

write_to_data(5,40)
```
