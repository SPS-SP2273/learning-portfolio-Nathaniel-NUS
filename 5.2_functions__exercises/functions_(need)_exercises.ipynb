### Exercise 1

```python
def greeting(name):
    if name == 'Batman':
        return 'Hello Batman! So, nice to meet you!'
    return f'Hello {name}!'

greeting('Joe')
```

### Exercise 2

```python
def calculate_bmi(weight, height):
    BMI = weight / (height ** 2)
    if BMI < 18.5:
        return "Underweight"
    elif 18.5 <= BMI <= 24.9:
        return "Normal weight"
    elif 25 <= BMI <= 29.9:
        return "Overweight"
    else:
        return "Obese"

calculate_bmi(80,1.76)
```

### Exercise 3

```python
def factorial(n):
    if n < 0:
        return "integer n cannot be a negative number"
    elif n == 0:
        return 1
    else:
        result = n
        # for loop will iterate from 1 to n-1. For each iteration, the number is multiplied to the result
        for i in range(1, n):
            result *= i
        return result

print(factorial(4))
print(factorial(-6))
print(factorial(0))
print(factorial(8))
```
