# Spherical coordinate system

- radius ($r$) has range of $r \geq 0$
- Polar angle ($\theta$) has range of $0 \leq \theta \leq \pi$
- Azimuthal angle ($\phi$) has range of $0 \leq \phi \leq 2\pi$

# Conversion from Spherical to Cartesian

- $z = r\cos\phi$
- $x = r\sin\phi\cos\theta$
- $y = r\sin\phi\sin\theta$

```python
from scipy.special import genlaguerre
from scipy.special import sph_harm_y
a = 52.9  #Bohr Radius

r = np.linspace(0, 3*a, 1000)

theta, phi = np.meshgrid(np.linspace(0, np.pi, 1000), np.linspace(0, 2 * np.pi, 1000))

def radial(n,l, r):
    rho = (2*r)/(n*a)
    R = np.sqrt(((2/(n*a))**3) * (math.factorial(n-l-1)/(2*n*math.factorial(n+l)))) * np.exp(-1*rho/2) * (rho**l) * genlaguerre(n-l-1,2*l+1)(rho)
    return R

def angular(m,l):
    Y_m = sph_harm_y(m, l, theta, phi)
    return Y_m

radial_part = radial(2,1,r) 
angular_part = angular(1,1)
wavefunction = radial_part * angular_part

e_density = np.abs(wavefunction)**2

x = np.sin(theta) * np.sin(phi)
y = np.sin(theta) * np.cos(phi) 

X, Y = x * radial_part * np.abs(angular_part), y * radial_part * np.abs(angular_part)

plt.contourf(X,Y,e_density)
plt.axis('equal')
plt.colorbar()
plt.show()
```
